```python
import pandas as pd
import duckdb
from pathlib import Path

# Define project paths
PROJECT_ROOT = Path.cwd().parent
trade_file = PROJECT_ROOT / "data" / "raw" / "FAOSTAT_data_en_matrix.csv"

print("--- LOADING TRADE DATA ---")
df_trade = pd.read_csv(trade_file)

print(f"Dataset Shape (Rows, Columns): {df_trade.shape}")

print("\nColumn Names:")
print(df_trade.columns.tolist())

print("\nFirst 3 Rows:")
display(df_trade.head(3))

print("\nData Types & Missing Values:")
df_trade.info()

print("\n--- TEST DUCKDB QUERY ---")

con = duckdb.connect(database=":memory:")
con.register("trade_df", df_trade)

result = con.execute("""
SELECT
    "Reporter Countries",
    COUNT(DISTINCT "Year") AS years_count,
    MIN("Year") AS min_yr,
    MAX("Year") AS max_yr
FROM trade_df
GROUP BY "Reporter Countries"
LIMIT 5;
""").fetchdf()

print(result)
```
